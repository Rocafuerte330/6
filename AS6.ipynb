## Punto 1 — Carga y limpieza de la base de datos

- Carga de base de datos.
- Eliminación las columnas: `Patient ID`, `Blood Pressure`, `Country`, `Continent`, `Hemisphere`. 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Carga de la base de datos
data = pd.read_csv('heart_attack_prediction_dataset.csv')

print('Dimensiones originales:', data.shape)
print('\nColumnas originales:')
print(data.columns.tolist())
```

```python
# Eliminación de las columnas indicadas en el enunciado
columnas_a_eliminar = ['Patient ID', 'Blood Pressure', 'Country', 'Continent', 'Hemisphere']
data = data.drop(columns=columnas_a_eliminar)

print('Dimensiones tras eliminar columnas:', data.shape)
print('\nColumnas restantes:')
print(data.columns.tolist())
```
